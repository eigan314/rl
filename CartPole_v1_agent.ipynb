### Custom Callback for Logging Training Metrics

To plot the loss curves, we need to collect the training metrics (such as `total_loss`, `policy_gradient_loss`, `value_loss`, `entropy_loss`, and `mean_episode_reward`) at each update step during the model's training process. This custom callback will store these metrics in a list, which can then be converted into a pandas DataFrame for plotting.

```python
import pandas as pd
from stable_baselines3.common.callbacks import BaseCallback

class TrainingLoggerCallback(BaseCallback):
    """
    A custom callback that logs training metrics (losses and episode rewards)
    from the Stable-Baselines3 logger during the training process.
    """
    def __init__(self, verbose=0):
        super(TrainingLoggerCallback, self).__init__(verbose)
        self.logs = []
        self._last_n_updates = -1 # To track when n_updates changes

    def _on_step(self) -> bool:
        # Access the internal logger of the model
        log_dict = self.logger.get_log_dict()

        # The 'n_updates' metric increments each time the policy is updated.
        # We want to log the metrics only once per update cycle.
        current_n_updates = log_dict.get('train/n_updates', -1)

        if current_n_updates > self._last_n_updates:
            # If n_updates has increased, new training metrics are available.
            # Append the current state of relevant metrics to our logs list.
            self.logs.append({
                'timesteps': self.num_timesteps,
                'total_loss': log_dict.get('train/loss', float('nan')),
                'policy_gradient_loss': log_dict.get('train/policy_gradient_loss', float('nan')),
                'value_loss': log_dict.get('train/value_loss', float('nan')),
                'entropy_loss': log_dict.get('train/entropy_loss', float('nan')),
                'ep_rew_mean': log_dict.get('rollout/ep_rew_mean', float('nan'))
            })
            self._last_n_updates = current_n_updates
        return True

    def get_training_data(self):
        """
        Returns the collected training data as a pandas DataFrame.
        """
        return pd.DataFrame(self.logs)
```

```python
from stable_baselines3 import PPO
from stable_baselines3.common.env_util import make_vec_env

# Instantiate the custom callback
training_logger = TrainingLoggerCallback()

vec_env = make_vec_env('CartPole-v1', n_envs=16)
model = PPO('MlpPolicy',
            vec_env,
            n_steps=64,
            batch_size=64,
            n_epochs=10,
            gamma=0.99,
            gae_lambda=0.98,
            ent_coef=0.01,
            verbose=1)
model.learn(total_timesteps=1000000, callback=training_logger) # Pass the callback here
model.save("ppo-cartpole-v1")
```

### Training Loss Curves

Now we can retrieve the logged training data from our custom callback and plot the evolution of the various loss functions (total loss, policy gradient loss, value loss, and entropy loss) over training timesteps. We'll also include the mean episode reward for context, to see how performance correlates with the losses.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Get the training data from the callback
training_df = training_logger.get_training_data()

if not training_df.empty:
    plt.figure(figsize=(15, 10))

    # Plot Total Loss
    plt.subplot(3, 2, 1)
    sns.lineplot(x='timesteps', y='total_loss', data=training_df)
    plt.title('Total Loss Over Training Time')
    plt.xlabel('Timesteps')
    plt.ylabel('Total Loss')
    plt.grid(True)

    # Plot Policy Gradient Loss
    plt.subplot(3, 2, 2)
    sns.lineplot(x='timesteps', y='policy_gradient_loss', data=training_df)
    plt.title('Policy Gradient Loss Over Training Time')
    plt.xlabel('Timesteps')
    plt.ylabel('Policy Gradient Loss')
    plt.grid(True)

    # Plot Value Loss
    plt.subplot(3, 2, 3)
    sns.lineplot(x='timesteps', y='value_loss', data=training_df)
    plt.title('Value Loss Over Training Time')
    plt.xlabel('Timesteps')
    plt.ylabel('Value Loss')
    plt.grid(True)

    # Plot Entropy Loss
    plt.subplot(3, 2, 4)
    sns.lineplot(x='timesteps', y='entropy_loss', data=training_df)
    plt.title('Entropy Loss Over Training Time')
    plt.xlabel('Timesteps')
    plt.ylabel('Entropy Loss')
    plt.grid(True)

    # Plot Mean Episode Reward
    plt.subplot(3, 1, 3)
    sns.lineplot(x='timesteps', y='ep_rew_mean', data=training_df)
    plt.title('Mean Episode Reward Over Training Time')
    plt.xlabel('Timesteps')
    plt.ylabel('Mean Episode Reward')
    plt.grid(True)

    plt.tight_layout()
    plt.show()
else:
    print("No training data collected. Please ensure the model was trained with the callback.")
```

# record video

```python
import gymnasium as gym
from stable_baselines3.common.vec_env import DummyVecEnv, VecVideoRecorder
from stable_baselines3.common.monitor import Monitor

# 1. Initialize the video wrapper environment
video_folder = "./videos"
video_length = 1000

eval_env = DummyVecEnv([lambda: Monitor(gym.make(env_id, render_mode="rgb_array"))])

# 2. Force it to record from step 0
eval_env = VecVideoRecorder(
    eval_env,
    video_folder,
    record_video_trigger=lambda x: x == 0,
    video_length=video_length,
    name_prefix="ppo-cartpole-v1"
)

# 3. Let the agent play the environment
obs = eval_env.reset()
for _ in range(video_length + 1):
    action, _states = model.predict(obs, deterministic=True)
    obs, _, _, _ = eval_env.step(action)

# Save and release the video file pointer
eval_env.close()
print("🎥 Video recording completed successfully!")
```
