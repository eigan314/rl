```python
import numpy as np
import gymnasium as gym
from gymnasium import spaces
```

```python
class LineWorldEnv(gym.Env):
  def __init__(self):
    super().__init__()

    self.size = 5   # 0 1 2 3 4
    self.goal = 4   # A _ _ _ G
    self.max_steps = 20

    self.action_space = spaces.Discrete(2)    # action space = either 0 (left) or 1 (right)
    self.observation_space = spaces.Discrete(self.size)   # observation space = total env space

  def reset(self, seed=None, options=None):
    super().reset(seed=seed)

    self.position = 0   # agent starts at pos 0
    self.steps = 0    # steps = 0

    observation = self.position # obs = what position the agent takes
    info = {}

    return observation, info

  def step(self, action):
    self.steps += 1     # keep incrementing steps

    if action == 0:
      self.position = -1    # if action = 0 (left), position = -1 (left)
    else:
      self.position = 1   #if action = 1 (right), position = +1 (right)

    self.position = np.clip(self.position, 0, self.size - 1)    #keep position between 0 and 4

    terminated = self.position == self.goal   # env is terminated if goal is achieved
    truncated = self.steps >= self.max_steps  # env is truncated if steps go beyond max_steps

    reward = 10 if terminated else -1  # reward = 10 if goal is achieved (terminated) else reward = -1

    observation = self.position
    info = {}

    return observation, reward, terminated, truncated, info

  def render(self):
    world = ["_"] * self.size
    world[self.position] = "A"
    world[self.goal] = "G"
    print(" ".join(world))

```

```python
env = LineWorldEnv()

obs, info = env.reset()
env.render()

for _ in range(10):
  action = env.action_space.sample()
  obs, reward, terminated, truncated, info = env.step(action)

  print("Action: ", action)
  print("Observation: ", obs)
  print("Reward: ", reward)
  print("Terminated: ", terminated)
  print("Truncated: ", truncated)
  print("Info: ", info)

  env.render()

  if terminated or truncated:
    break
```
